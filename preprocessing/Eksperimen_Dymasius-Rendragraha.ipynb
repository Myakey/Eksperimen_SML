```python
# Mendefinisikan path dataset sesuai instruksi kamu
train_path = '../airplane_satisfaction_raw/train.csv'
test_path = '../airplane_satisfaction_raw/test.csv'

# Memuat dataset
df_train = pd.read_csv(train_path)
df_test = pd.read_csv(test_path)

print("Dimensi Data Train:", df_train.shape)
print("Dimensi Data Test:", df_test.shape)
```

# **5. Data Preprocessing**

Pada tahap ini, data preprocessing adalah langkah penting untuk memastikan kualitas data sebelum digunakan dalam model machine learning.

Jika Anda menggunakan data teks, data mentah sering kali mengandung nilai kosong, duplikasi, atau rentang nilai yang tidak konsisten, yang dapat memengaruhi kinerja model. Oleh karena itu, proses ini bertujuan untuk membersihkan dan mempersiapkan data agar analisis berjalan optimal.

Berikut adalah tahapan-tahapan yang bisa dilakukan, tetapi **tidak terbatas** pada:
1. Menghapus atau Menangani Data Kosong (Missing Values)
2. Menghapus Data Duplikat
3. Normalisasi atau Standarisasi Fitur
4. Deteksi dan Penanganan Outlier
5. Encoding Data Kategorikal
6. Binning (Pengelompokan Data)

Cukup sesuaikan dengan karakteristik data yang kamu gunakan yah. Khususnya ketika kami menggunakan data tidak terstruktur.

```python
# 1. Menghapus kolom yang tidak relevan (Unnamed: 0 dan id)
cols_to_drop = ['Unnamed: 0', 'id']
df_train = df_train.drop(columns=[col for col in cols_to_drop if col in df_train.columns], errors='ignore')
df_test = df_test.drop(columns=[col for col in cols_to_drop if col in df_test.columns], errors='ignore')

# 2. Menangani Missing Values dengan menghindari Data Leakage
# Ada 310 missing values di 'Arrival Delay in Minutes'.
# Kita hitung median dari df_train, dan gunakan untuk mengisi df_train maupun df_test
median_delay = df_train['Arrival Delay in Minutes'].median()
df_train['Arrival Delay in Minutes'] = df_train['Arrival Delay in Minutes'].fillna(median_delay)
df_test['Arrival Delay in Minutes'] = df_test['Arrival Delay in Minutes'].fillna(median_delay)

# 3. Encoding Data Kategorikal
# a. Encoding tipe Ordinal (Ada tingkatannya) untuk 'Class'
class_mapping = {'Eco': 0, 'Eco Plus': 1, 'Business': 2}
if 'Class' in df_train.columns:
    df_train['Class'] = df_train['Class'].map(class_mapping)
    df_test['Class'] = df_test['Class'].map(class_mapping)

# b. Label Encoding untuk tipe Nominal/Binary dan Target ('satisfaction')
categorical_cols = ['Gender', 'Customer Type', 'Type of Travel', 'satisfaction']
le = LabelEncoder()
for col in categorical_cols:
    if col in df_train.columns:
        # Fit di train, lalu transform di train dan test
        df_train[col] = le.fit_transform(df_train[col])
        df_test[col] = le.transform(df_test[col])

# 4. Standarisasi Fitur Numerik Kontinu
# Catatan: Fitur rating seperti 'Inflight wifi service' (int64) tidak di-scale agar 
# skalanya tetap asli (misal 0-5), sehingga lebih mudah diinterpretasi model.
num_cols = ['Age', 'Flight Distance', 'Departure Delay in Minutes', 'Arrival Delay in Minutes']
scaler = StandardScaler()
df_train[num_cols] = scaler.fit_transform(df_train[num_cols])
df_test[num_cols] = scaler.transform(df_test[num_cols]) # Pakai scaler hasil fit dari data train

# 5. Menyimpan Hasil Preprocessing ke Direktori Tujuan
output_dir = './preprocessing/airplane_satisfaction_preprocessing'
os.makedirs(output_dir, exist_ok=True)

df_train.to_csv(f'{output_dir}/train_preprocessed.csv', index=False)
df_test.to_csv(f'{output_dir}/test_preprocessed.csv', index=False)

print(f"Sisa missing values di train: {df_train.isnull().sum().max()}")
print(f"Sisa missing values di test: {df_test.isnull().sum().max()}")
print(f"Proses selesai! Data telah berhasil disave di: {output_dir}")
```
